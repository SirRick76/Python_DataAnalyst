# While

```python
#pedir ao utilizador uma idade. Enquanto idade for menor que 0 e maior que 100 voltar a pedir idade
idade = 0
while idade <= 0 or idade > 100:
    idade = int(input("Por favor, insira a sua idade (entre 0 e 100): "))
print(f"A sua idade é: {idade}")
```

```python
idade = None
while idade is None or idade < 0 or idade > 100:
    idade = int(input("Por favor, insira a sua idade (entre 0 e 100): "))
print(f"A sua idade é: {idade}")
```

```python
# simular um repita
valida = False
while not valida:
    idade = int(input("Por favor, insira a sua idade (entre 0 e 100): "))
    if idade>=0 and idade<=100:
        valida = True
print(f"A sua idade é: {idade}")
```

```python
# com break
while True:
    idade = int(input("Por favor, insira a sua idade (entre 0 e 100): "))
    if idade>=0 and idade<=100:
        print("Idade valida!")
        break
    print("Idade invalida! Tente novamente.")
print(f"A sua idade é: {idade}")
```

# FOR

```python
letras = ['a','b','c','d', 'e'] # lista de letras
for letra in letras: # para cada letra na lista de letras
    print(letra) 
```

```python
for i in range(5):
    print(i)
print('----------------------')
for i in range(3, 7):
    print(i)
print('----------------------')
for i in range(10, 21, 3):
    print(i)
```
