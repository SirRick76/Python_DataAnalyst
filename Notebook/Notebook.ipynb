# Sequencia de Caracteres 

```python
texto="Python"
print(texto[0])

texto= "J"+texto[1:]
print(texto)
```

```python
codigo="pt-2026-brg"
print(codigo[0:2])
print(codigo[3:7])
print(codigo[8:11])
print(codigo[0:2].upper()+codigo[2:11])
# strip()-> retira espaços em branco do início e do fim da string
codigo="  pt-2026-brg  "
print(codigo.strip())
frase="Curso cesae data analyst"
print(frase.split())
print(type(frase.split()))
```
