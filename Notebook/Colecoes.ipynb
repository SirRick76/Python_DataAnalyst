# Listas

```python
cidade = ["Braga","Porto","Lisboa"]
print (cidade)
print (cidade[0])
print (cidade[1])
print (cidade[2])

lista = [20, 31, 42, 53, 64, 55]
print(lista[2])
lista[2]=100
print(lista[2])
tamanho_lista=len(lista)
print(tamanho_lista)
minimo_lista=min(lista)
print(minimo_lista)
maximo_lista=max(lista)
print(maximo_lista)
somatorio_lista=sum(lista)
print(somatorio_lista)
sorted_lista=sorted(lista)
print(sorted_lista)
print(31 in lista)
print(51 in lista)

L=["a","b","c","d","e"]
L[1:2]=[1,2,3]
print(L)
```
