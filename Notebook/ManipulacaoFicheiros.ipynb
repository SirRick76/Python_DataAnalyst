Ler info ficheiro notas_ficheiro_teste

```python
alunos=[]
with open("notas_ficheiro_teste.txt","r",encoding="utf-8") as ficheiro:
    #conteudo = ficheiro.read()
    lista = ficheiro.readlines()
    for linha in lista:
        dados = linha.strip().split(";")
        alunos.append(dados)

print("Lista de alunos:")
print(alunos)

#aceder a informação da Ana
print(alunos[1][0])
print(alunos[1][1])
print(alunos[1][2])

print("_"*20)

# quais os alunos com a nota superior a 15
alunos_nota_superior_15 = []
for aluno in alunos[1:]:
    if float(aluno[2]) > 15:
        alunos_nota_superior_15.append(aluno)

print("Alunos com nota superior a 15:")
print(alunos_nota_superior_15)
```
